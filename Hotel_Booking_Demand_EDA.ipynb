```python
# Hotel Booking Demand — Exploratory Data Analysis

## 1. Introduction

This project focuses on performing Exploratory Data Analysis (EDA) on the Hotel Booking Demand dataset.

The dataset contains hotel reservation information from a city hotel and a resort hotel. It includes booking details, customer information, reservation dates, pricing information, and cancellation status.

The objective is to understand booking patterns, identify missing data, investigate relationships between variables, and discover meaningful business insights using Python.

## 2. Objectives

- Understand the structure and characteristics of the dataset.
- Identify missing values and duplicate records.
- Perform descriptive statistical analysis.
- Explore numerical and categorical variables.
- Visualize booking trends and relationships.
- Identify patterns associated with reservation cancellations.
- Summarize key findings based on the analysis.

## 3. Tools and Technologies

- Python
- Pandas
- NumPy
- Matplotlib
- Seaborn
- Google Colab
```

```python
# Import essential libraries

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configure visualization settings
sns.set_theme(style="whitegrid")

plt.rcParams['figure.figsize'] = (10, 6)
plt.rcParams['font.size'] = 11

# Display all columns
pd.set_option('display.max_columns', None)

print("Libraries imported successfully!")
```

```python
# Load dataset

df = pd.read_csv('/hotel_bookings.csv')

# Display first five records
df.head()
```

```python
# Count missing values in each column

missing_values = df.isnull().sum()

print(missing_values[missing_values > 0])
```

```python
# Calculate missing value percentage

missing_percentage = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    'Missing Values': df.isnull().sum(),
    'Missing Percentage': missing_percentage
})

missing_summary = missing_summary[
    missing_summary['Missing Values'] > 0
].sort_values(by='Missing Percentage', ascending=False)

missing_summary
```

```python
plt.figure(figsize=(12, 6))

missing_summary['Missing Percentage'].plot(
    kind='bar',
    color='steelblue'
)

plt.title('Missing Values by Column')
plt.xlabel('Columns')
plt.ylabel('Missing Values (%)')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Count duplicate rows

duplicate_count = df.duplicated().sum()

print("Duplicate Records:", duplicate_count)
```

```python
duplicate_percentage = (duplicate_count / len(df)) * 100

print("Duplicate Percentage:", duplicate_percentage, "%")
```

```python
df[df.duplicated()].head()
```

```python
# Preserve original data

df_clean = df.copy()

# Remove exact duplicate records

df_clean = df_clean.drop_duplicates()

print("Original Shape:", df.shape)
print("Cleaned Shape:", df_clean.shape)
```

```python
df_clean.describe()
```

```python
df_clean.describe(include='object')
```

```python
categorical_columns = [
    'hotel',
    'meal',
    'market_segment',
    'distribution_channel',
    'customer_type',
    'deposit_type',
    'reservation_status'
]

for column in categorical_columns:
    print("\n", column)
    print(df_clean[column].value_counts(dropna=False))
```

```python
print("Negative ADR:", (df_clean['adr'] < 0).sum())

print("Negative Adults:", (df_clean['adults'] < 0).sum())

print("Negative Children:", (df_clean['children'] < 0).sum())

print("Zero Guest Bookings:",
      ((df_clean['adults'] +
        df_clean['children'].fillna(0) +
        df_clean['babies']) == 0).sum())
```

```python
# Descriptive statistics

df_clean.describe()
```

```python
numeric_summary = pd.DataFrame({
    'Mean': df_clean.mean(numeric_only=True),
    'Median': df_clean.median(numeric_only=True),
    'Standard Deviation': df_clean.std(numeric_only=True)
})

numeric_summary
```

```python
Q1 = df_clean['adr'].quantile(0.25)
Q3 = df_clean['adr'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df_clean[
    (df_clean['adr'] < lower_bound) |
    (df_clean['adr'] > upper_bound)
]

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of Outliers:", len(outliers))
```

```python
plt.figure(figsize=(10, 6))

sns.histplot(
    data=df_clean,
    x='lead_time',
    bins=50,
    kde=True
)

plt.title('Distribution of Hotel Booking Lead Time')
plt.xlabel('Lead Time (Days)')
plt.ylabel('Number of Bookings')

plt.tight_layout()
plt.show()
```

```python
cancellation_counts = df_clean['is_canceled'].value_counts()

plt.figure(figsize=(8, 5))

sns.barplot(
    x=['Not Cancelled', 'Cancelled'],
    y=[
        cancellation_counts.get(0, 0),
        cancellation_counts.get(1, 0)
    ]
)

plt.title('Hotel Booking Cancellation Distribution')
plt.xlabel('Booking Status')
plt.ylabel('Number of Bookings')

plt.tight_layout()
plt.show()
```

```python
cancellation_percentage = (
    df_clean['is_canceled'].value_counts(normalize=True) * 100
)

print(cancellation_percentage)
```

```python
# Convert arrival month into a consistent monthly ordering

month_order = [
    'January', 'February', 'March', 'April',
    'May', 'June', 'July', 'August',
    'September', 'October', 'November', 'December'
]

df_clean['arrival_date_month'] = pd.Categorical(
    df_clean['arrival_date_month'],
    categories=month_order,
    ordered=True
)
```

```python
monthly_bookings = (
    df_clean.groupby('arrival_date_month', observed=False)
    .size()
    .reset_index(name='Bookings')
)
```

```python
plt.figure(figsize=(12, 6))

sns.lineplot(
    data=monthly_bookings,
    x='arrival_date_month',
    y='Bookings',
    marker='o'
)

plt.title('Monthly Hotel Booking Trends')
plt.xlabel('Month')
plt.ylabel('Number of Bookings')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
numeric_data = df_clean.select_dtypes(
    include=['int64', 'float64']
)
```

```python
plt.figure(figsize=(9, 6))

sns.boxplot(
    data=df_clean,
    x='hotel',
    y='adr'
)

plt.title('Average Daily Rate by Hotel Type')
plt.xlabel('Hotel Type')
plt.ylabel('Average Daily Rate')

plt.tight_layout()
plt.show()
```

```python
top_countries = (
    df_clean['country']
    .value_counts()
    .head(10)
    .reset_index()
)

top_countries.columns = ['Country', 'Bookings']
```

```python
plt.figure(figsize=(10, 6))

sns.barplot(
    data=top_countries,
    x='Bookings',
    y='Country'
)

plt.title('Top 10 Countries by Number of Hotel Bookings')
plt.xlabel('Number of Bookings')
plt.ylabel('Country')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

sns.countplot(
    data=df_clean,
    x='market_segment',
    order=df_clean['market_segment'].value_counts().index
)

plt.title('Hotel Bookings by Market Segment')
plt.xlabel('Market Segment')
plt.ylabel('Number of Bookings')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

sns.countplot(
    data=df_clean,
    x='market_segment',
    order=df_clean['market_segment'].value_counts().index
)

plt.title('Hotel Bookings by Market Segment')
plt.xlabel('Market Segment')
plt.ylabel('Number of Bookings')
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

sns.boxplot(
    data=df_clean,
    x='is_canceled',
    y='lead_time'
)

plt.title('Lead Time vs Booking Cancellation')
plt.xlabel('Cancellation Status (0 = No, 1 = Yes)')
plt.ylabel('Lead Time (Days)')

plt.tight_layout()
plt.show()
```

```python
cancellation_by_hotel = (
    df_clean.groupby('hotel', observed=True)['is_canceled']
    .mean() * 100
)

print(cancellation_by_hotel)
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(
    x=cancellation_by_hotel.index,
    y=cancellation_by_hotel.values
)

plt.title('Cancellation Rate by Hotel Type')
plt.xlabel('Hotel Type')
plt.ylabel('Cancellation Rate (%)')

plt.tight_layout()
plt.show()
```

```python
customer_cancellation = (
    df_clean.groupby('customer_type', observed=True)['is_canceled']
    .mean() * 100
)

print(customer_cancellation)
```

```python
plt.figure(figsize=(10, 5))

sns.barplot(
    x=customer_cancellation.index,
    y=customer_cancellation.values
)

plt.title('Cancellation Rate by Customer Type')
plt.xlabel('Customer Type')
plt.ylabel('Cancellation Rate (%)')

plt.tight_layout()
plt.show()
```
